Day 7

```python


print("==========================================")
print("   AI DOCUMENT QUESTION-ANSWERING APP")
print("==========================================")

document = """
Artificial Intelligence is a technology that enables
computers to perform intelligent tasks.

Machine Learning is a part of Artificial Intelligence.
It allows computers to learn from data.

Generative AI can create new content such as text,
images, code and answers.

Large Language Models can understand and generate text.
RAG retrieves useful information before generating an answer.
"""

print("\nDocument Loaded Successfully!")

chunks = document.split("\n\n")

print("\nDocument Chunks:")
for i, chunk in enumerate(chunks, 1):
    print("\nChunk", i)
    print(chunk)
questions = [
    "What is Artificial Intelligence?",
    "What is Machine Learning?",
    "What is Generative AI?",
    "What is LLM?",
    "What is RAG?"
]

answers = [
    "Artificial Intelligence enables computers to perform intelligent tasks.",
    "Machine Learning allows computers to learn from data.",
    "Generative AI can create new content such as text, images and code.",
    "Large Language Models can understand and generate text.",
    "RAG retrieves useful information before generating an answer."
]

print("\nAvailable Questions:")
for i, q in enumerate(questions, 1):
    print(i, ".", q)


choice = 3

print("\nSelected Question:")
print(questions[choice - 1])
print("\nRetrieved Information:")
print(answers[choice - 1])


print("\nAI Answer:")
print(answers[choice - 1])

print("\n==========================================")
print("   DAY 7 COMPLETED SUCCESSFULLY!")
print("==========================================")
```
